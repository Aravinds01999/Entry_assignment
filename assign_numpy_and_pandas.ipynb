```python
import numpy as  np
import pandas as pd
```

```python
#1. Create a numpy array containing the numbers from 1 to 10, and then reshape it to a 2x5 matrix. (Score:1) 

a = np.arange(1,11)
mat = a.reshape(2, 5)
print(mat)
```

```python
#2. Create a numpy array containing the numbers from 1 to 20, and then extract the elements between the 5th and 15th index. (Score:1) 

b = np.arange(1, 21)
b1 = b[5:16]
print(b1)
```

```python
#3. How to compute the mean, median, and standard deviation of a numpy array?Use the array you created above (Score 1)

mean = np.mean(b)
median = np.median(b)
sd = np.std(b)
print(mean)
print(median)
print(sd)
```

```python
#4. Write a NumPy program that creates a 2D array x of shape (3, 4) and a 1D array y of shape (4,). Subtract y from each row of x using broadcasting. (Score: 2)  

x = np.arange(12).reshape(3, 4)
print(x)   

y = np.array([1, 2, 3, 4])   
print(y)

z = x - y
print(z)
```

```python
"""5. Create a dataframe with the following columns: name, age, and gender. The dataframe 
should have 10 rows of data. (Score: 2) 
 
"""


employe = {
    "name": ["Amal", "Jacob", "Sam", "Tony", "Vinu", "Maria", "John", "Hari", "Sunil", "Jishnu"],
    "age": [25, 32, 28, 40, 35, 22, 30, 27, 45, 29],
    "gender": ["M", "F", "M", "F", "M", "F", "M", "F", "M", "F"]
}

df = pd.DataFrame(employe)
print(df)
```

```python
#1) Add a new column to the data frame created in question 1, called occupation. The values for this column should be Programmer, Manager, and Analyst, corresponding to the rows in the dataframe. (Score: 1) 

df["occupation"] = ["Programmer", "Manager", "Analyst", "Programmer", 
                    "Manager", "Analyst", "Programmer", "Manager", 
                    "Analyst", "Programmer"]
print(df)
```

```python
#2) Select the rows of the dataframe where the age is greater than or equal to 30. (Score: 1)

age_filter = df[df["age"] >= 30]
print(age_filter)
```

```python
#3) Convert this dataframe to a CSV file and read that CSV file, and finally display the contents. (Score: 1)

df.to_csv("company.csv", index=False)   
new_df = pd.read_csv("company.csv")     
print(new_df)
```
