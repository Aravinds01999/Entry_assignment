```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
#1. Load the dataset and handle any missing data by replacing null values with the meanvalue of the respective column.

df = pd.read_csv("winequality-red.csv", sep =";")
df.head()                 
```

```python

df.fillna(df.mean(numeric_only=True), inplace = True)
df.isnull().sum()

```

```python
#2. Extract the following columns as vectors: alcohol,citric acid.
alcohol_vec = df["alcohol"].values
citric_vec = df["citric acid"].values
[("Alcohol first 5:", alcohol_vec[:5]),
("Citric acid first 5:", citric_vec[:5])]
```

```python
#3. Select two features (e.g., alcohol and density) from the dataset and calculate thecovariance matrix using np.cov(X.T), where X is the feature matrix consisting of theselected columns.

X = df[['alcohol', 'density']].values
cov_matrix = np.cov(X.T)
cov_matrix
```

```python
#4. Perform eigen decomposition on the covariance matrix you computed in question 3.Identify and interpret the results:Identify the top 2 eigenvalues of the covariancematrix,Identify the corresponding eigenvectors.

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigen Value: \n", eigenvalues)
print("Eigen Vectors: \n", eigenvectors)
```

```python
#Which wine quality is most common in the dataset? How can you interpret thedistribution of wine quality scores?

quality_counts = df['quality'].value_counts()
most_common_quality = quality_counts.idxmax()

print("Most common wine quality:", most_common_quality)
print("Distribution:\n", quality_counts)
```
