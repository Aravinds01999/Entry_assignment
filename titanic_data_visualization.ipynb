```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
```

```python
"""1. Countplot: Passenger Count by Embarked Port  
● Create a countplot to display the number of passengers who boarded from each port (embarked), separated by sex."""


sns.countplot(data=titanic, x="embarked", hue="sex", palette="Set2")
plt.title("Passenger Count by Embarked Port and Sex")
plt.xlabel("Embarked Port")
plt.ylabel("Passenger Count")
plt.legend(title="Sex")
plt.show()
```

```python
Observation: Most boarded at S, with more males than females.
```

```python
"""2. Swarmplot: Age Distribution across Classes  
● Create a swarmplot to visualize the distribution of age across passenger classes (pclass), categorized by sex."""


plt.figure(figsize=(15,8))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex", palette="Set1", dodge=True)
plt.title("Age Distribution across Passenger Classes by Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()
```

```python
Observation:=- Firstt class spans a wider age range, while third class has more younger males.
```

```python
"""3. Pie Chart: Gender Distribution  
● Create a pie chart using Matplotlib to show the percentage of male vs. female passengers. - Add percentage labels using autopct - Use explode for one slice to 
highlight it."""

gender_counts = titanic['sex'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', explode=[0.1,0], colors=["lightblue","lightpink"], startangle=90)
plt.title("Gender Distribution of Passengers")
plt.show()
```

```python
Observation:- Around 65% male and 35% female.
```

```python
"""4. Heatmap: Correlation Matrix  
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch, and survived. (Use seaborn.heatmap with annotations)"""

corr = titanic[['age','fare','pclass','sibsp','parch','survived']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Titanic Features")
plt.show()
```

```python
Observation:-  Fare and class are strongly correlated, survival links weakly with fare/class.
```

```python
"""5. Violin Plot: Fare by Class and Gender  
● Create a violin plot to visualize fare distribution across classes, separated by gender"""

plt.figure(figsize=(10,6))
sns.violinplot(data=titanic, x="pclass", y="fare", hue="sex", split=True, palette="muted")
plt.title("Fare Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()
```

```python
Observation:- 1st class fares are highest; females often paid slightly more.
```

```python
"""6. Scatter Plot: Age vs Fare  
● Create a scatter plot to explore the relationship between age and fare, colored by 
survival status.  
● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels, 
and legends where needed.  """


plt.figure(figsize=(10,6))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived", palette="coolwarm", alpha=0.7)
plt.title("Age vs Fare Colored by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived", labels=["No","Yes"])
plt.show()
```

```python
Observation:- Higher fares correlate with better survival chances, especially among younger passengers.
```
